## Advance Example with Business Examples

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data Structure: [resturant_id,2021,2022,2023,2024]

sales_data=np.array([
    [1, 150000, 180000, 220000, 250000], #Prasid biryani
    [2, 120000, 145000, 190000, 230000], #Bikut
    [3, 200000, 225000, 260000, 300000], #Pizza
    [4, 95000, 125000, 160000, 185000],  #Burger
    [5, 175000, 210000, 245000, 280000], #Momo
])

print("===Zomato Salse Analysis===")
print("\nsalse data shape ",sales_data.shape)
print("sample data from first three restaurant:\n",sales_data[0:3])
print("sample data from first three restaurant:\n",sales_data[:,1:])
```

```python
#Total salse per year

yearly_total_row_wise=np.sum(sales_data[:,1:],axis=1)
print("sum data row",yearly_total_row_wise)
grand_total = np.sum(yearly_total_row_wise)
print("sum", grand_total)
yearly_total_column_wise=np.sum(sales_data[:,1:],axis=0)
print("sum data column",yearly_total_column_wise)
grand_total = np.sum(yearly_total_column_wise)
print("sum", grand_total)

#Minium salse per restaurant
min_sales=np.min(sales_data[:,1:],axis=1)
print("Min_salse row wise",min_sales)

# Maximum salse per year
max_sales=np.max(sales_data[:,1:],axis=0)
print("Min_salse column wise",max_sales)

# Average salse per restaurant
Average_Salse=np.mean(sales_data[:,1:],axis=1) #mean or average me kuch bhi use kr skte hai
print("Average_Salse",Average_Salse)
```

```python
#cumsum salse
cumsum=np.cumsum(sales_data[:,1:],axis=1)
print("Cumsum salse",cumsum)


#figar banana hai
plt.figure(figsize=(10,6))
plt.plot(np.mean(cumsum,axis=0))
plt.title("Average cumulative salse accross all resuaurant")
plt.xlabel("Years")
plt.ylabel("Salse")
plt.grid(True)
plt.show()
```

### vector

```python
vector1=np.array([1,2,3,4,5,6])
vector2=np.array([6,7,8,9,10,11])

print("Vector addition: ",vector1+vector2)
print("VEctor Multilication: ",vector1*vector2)
print("Dot Product",np.dot(vector1,vector2))

Angle=np.arccos(np.dot(vector1,vector2)/(np.linalg.norm(vector1)*np.linalg.norm(vector2)))
print("Angle",Angle)
```
