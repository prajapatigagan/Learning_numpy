## numpy array and Basics

```python
import numpy as np
```

### Creating array from list

```python
arr_1d=np.array([1,2,3,4,5,6])
print("1D Array",arr_1d)

arr_2d=np.array([[1,2,3],[5,6,3]])
print("2D Array",arr_2d)
```

### List vs Numpy array

```python
py_list=[1,2,3]
print("python list multiplication",py_list*2)

np_arr=np.array([1,2,3]) #Element wise multiplication
print("python array multiplication",np_arr*2)

import time

start=time.time()
py_list=[i*2 for i in range(10000)]
print("\n List Operation Time: ",time.time()-start)

start=time.time()
np_arr=np.arange(10000)*2
print("\n Numpy Operation Time: ",time.time()-start)
```

### Creating array from scratch

```python
zeros=np.zeros((3,4))
print("Array \n",zeros)

ones=np.ones((2,3))
print("Array \n",ones)

full=np.full((3,3),7)
print("Array \n",full)

random=np.random.random((2,3))
print("Array\n",random)

sequence=np.arange(0,20,2)
print("Arrays \n",sequence)
```

### Vector, Matrix and Tensor

```python
vector=np.array([1,2,3])
print("Vector\n",vector)

matrix=np.matrix([[1,2,3],[5,2,3],[4,2,3]])
print("Matrix\n",matrix)

tensor=np.array([[[1, 2],[3, 4]],[[3,5],[6,7]]])
print("Tensor\n",tensor)
```

### Arrays Properties

```python
arr=np.array([[1,2,3],[5,2,3],[4,2,3]])
print("Shape ",arr.shape)
print("Dimension ",arr.ndim)
print("Size ",arr.size)
print("Dtype ",arr.dtype)
```

### Array Reshaping

```python
arr=np.arange(16)
print("Original Array \n",arr)

reshaped=arr.reshape((4,4))
print("Reshap Array\n",reshaped)

flattend=reshaped.flatten()
print("Flattend Array \n",flattend)

# rave(returns view, instead of copy)
raveled=reshaped.ravel()
print("Raveled Array\n",raveled)

#Transpose
transpose=reshaped.T
print("Transpose\n",transpose)
```
